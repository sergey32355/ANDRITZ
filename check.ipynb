```python
import os
import pickle
import matplotlib.pyplot as plt
```

```python
files = os.listdir("data/RT/")
n_files = len(files)
fig, ax = plt.subplots(n_files, 1, figsize=(12, 3 * n_files))

if n_files == 1:
    ax = [ax]

for idx, file in enumerate(files):
    with open(f"data/RT/{file}", "rb") as f:
        data = pickle.load(f)
    data.pop(list(data.keys())[-1])
    for name, channel in list(data.items())[:3]:
        ax[idx].plot(channel[::100], label=name, alpha=0.5)
    ax[idx].legend(loc="upper right")

plt.tight_layout()
```
